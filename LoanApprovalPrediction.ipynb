# **OneHotEncoding **

```python
from sklearn.preprocessing import OneHotEncoder

object_cols =['Gender' ,  'Married' , 'Education','Self_Employed' ,  'Property_Area']
print(object_cols)
OH_Encoder = OneHotEncoder(sparse_output=False , handle_unknown='ignore')

OH_Cols = pd.DataFrame(OH_Encoder.fit_transform(df[object_cols]))

OH_Cols.index = df.index

OH_Cols.columns = OH_Encoder.get_feature_names_out()

df_final = df.drop(object_cols , axis=1)
df_final = pd.concat([df_final , OH_Cols] , axis = 1)

df_final.info()
```

# ***split data => tranning data testing data***

```python
from sklearn.model_selection import train_test_split

x  = df_final.drop(['Loan_ID', 'Loan_Status'] , axis= 1)
y = df_final['Loan_Status']

x_train , x_test , y_train , y_test = train_test_split(x , y , random_state=42 , test_size=0.2 )

x_train.shape , x_test.shape , y_train.shape , y_test.shape
```

***create the model***

```python
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier

lg_model = LogisticRegression()
lg_model.fit(x_train , y_train)
y_pred_lg = lg_model.predict(x_test)

rfc_model = RandomForestClassifier(n_estimators=7 , criterion='entropy' , random_state=7)
rfc_model.fit(x_train , y_train)
y_pred_rfc = rfc_model.predict(x_test)

```

# Acuurecy => Model Performance

```python
from sklearn.metrics import accuracy_score

acc_lg = accuracy_score(y_test , y_pred_lg)
acc_rfc = accuracy_score(y_test , y_pred_rfc)

acc_lg  # 0.801980198019802

acc_rfc  # 0.8316831683168316

# RandomForestClassifier the best model
```
